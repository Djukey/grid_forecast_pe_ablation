```
# Compare installed libraries with the pinned requirements.txt (a mismatch is reported, not hidden)
import importlib.metadata as md
for line in open('requirements.txt'):
    name, want = line.strip().split('==')
    have = md.version(name).split('+')[0]
    print(f"{name:12s} pinned {want:8s} installed {have:8s}", 'OK' if have == want else 'MISMATCH')
```
